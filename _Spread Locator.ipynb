1. What is Statistical Distributions?
A statistical distribution is a way of describing how different values of a dataset are spread and how frequently they occur. It helps us understand the behaviour and pattern of data. In statistics, we use distributions to study the possible values of a random variable and the probability associated with those values.
Whenever we collect data all the values are not usually the same. Some values occur more frequently, while some values occur less frequently. A statistical distribution helps us organise and understand these differences.

2. What is a Q-Q Plot and Why is it Used?
A Q-Q Plot stands for Quantile-Quantile Plot. It is a graphical technique used in statistics to compare the distribution of sample data with a theoretical distribution.
A Q-Q plot helps us understand whether the given data follows a particular distribution, such as a normal distribution. It compares the quantiles of actual sample data with the quantiles of a theoretical reference distribution.
In a Q-Q plot, the theoretical quantiles are generally represented on the X-axis, while the sample quantiles are represented on the Y-axis. If the sample data follows the selected theoretical distribution, most of the plotted points should lie approximately along a straight reference line.
A Q-Q plot is mainly used to check whether data follows a normal distribution. It is also useful for identifying outliers, understanding skewness and examining the behaviour of the tails.
Q-Q plot is a diagnostic tool, not proof of normality. A straight line does not guarantee that the data is perfectly normally distributed, especially when the sample size is small.

3. Difference Between Discrete and Continuous Distributions
1 Discrete Distribution
A discrete distribution is used when a random variable can take separate and countable values. These values may be finite or countably infinite. There are gaps between the possible values.
Discrete distributions are mainly used for counting things, such as the number of students in a classroom, the number of customers visiting a shop, the number of defective products and the number of heads obtained when tossing a coin.
For discrete random variables, we use the Probability Mass Function It gives the probability of each individual possible value.
ex-when we toss a coin three times, the possible number of heads is 0, 1, 2 or 3. These are countable values.

2 Continuous Distribution
A continuous distribution is used when a random variable can take any value within a given interval. There are no gaps between the possible values.
Continuous distributions are mainly used for measurements such as height, weight, temperature, time and delivery duration.
For continuous random variables, we use the Probability Density Function. The probability of a variable falling within an interval is calculated by finding the area under the density curve.
ex-the height of a student can be 160 cm, 160.5 cm, 160.55 cm or many other values.

4. What is Bernoulli Distribution?
Bernoulli Distribution is a discrete probability distribution that describes a random experiment having only two possible outcomes. These two outcomes are generally called success and failure.
In a Bernoulli experiment, success is represented by 1 and failure is represented by 0. The experiment has only one trial, and the probability of success is represented by p. Similarly, the probability of failure is represented by 1-p.

5. What is Binomial Distribution?
Binomial Distribution is a discrete probability distribution used to calculate the probability of obtaining a particular number of successes in a fixed number of independent trials.
It is an extension of the Bernoulli distribution. In Bernoulli distribution, we perform only one trial, whereas in Binomial distribution, we perform multiple independent trials.
Each trial has only two possible outcomes: success or failure. The probability of success remains constant for every trial, and the trials must be independent of each other

6. Explain Log-Normal Distribution.
Log-Normal Distribution is a continuous probability distribution used when the natural logarithm of a random variable follows a normal distribution.
 when we take the natural logarithm of a positive variable and the resulting values follow a normal distribution, the original variable is called log-normally distributed.
Log-Normal distribution is different from normal distribution because it is generally right-skewed and contains only positive values.

7. Explain Power Law Distribution.
Power Law Distribution is a type of probability distribution used to describe situations where a small number of observations have very large values, while most observations have comparatively small values.
It is mainly associated with data that has a heavy tail. A heavy tail means that extreme values occur more frequently than we would normally expect in a light-tailed distribution such as the normal distribution.
In a power-law pattern, the probability decreases as the value of the variable increases, but it decreases according to a power of that value rather than an exponential rate.

8. What is Box-Cox Transform?
Box-Cox Transformation is a statistical transformation technique used to change the shape of data and make it more suitable for statistical analysis.
In real-world datasets, the data is not always normally distributed. Sometimes the data is highly skewed, which can affect statistical methods that assume a particular distribution.
Box-Cox transformation is a data-driven power transformation. It helps us find a suitable power of the original data so that the transformed data becomes more symmetric and closer to a normal distribution

10. What is Z-Score Probability?
A Z-score is a statistical measurement that tells us how far a particular observation is from the mean of a distribution in terms of standard deviations.
It is also called a standardised value. It helps us understand whether an observation is below the average, equal to the average or above the average.
Z-scores are commonly used with the normal distribution. By converting the original values into Z-scores, we can use the standard normal distribution to calculate probabilities.

Formula of Z-Score
Z={X-u}/{sigma}

11. Differentiate Probability Density Function (PDF) and Cumulative Distribution Function (CDF).
1 Probability Density Function (PDF)
PDF stands for Probability Density Function. It is mainly used for continuous random variables and describes how probability is distributed over different values.
The PDF does not directly give the probability of an exact value. Instead, it gives the density of probability at a particular point.
To calculate the probability that a continuous random variable falls within an interval, we calculate the area under the PDF curve between the two specified values

2 CDF stands for Cumulative Distribution Function. It gives the accumulated probability that a random variable is less than or equal to a particular value.
it tells us how much probability has accumulated from the beginning of the distribution up to a given point.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from scipy.stats import bernoulli, binom, poisson, lognorm
from scipy.stats import boxcox
from statsmodels.graphics.gofplots import qqplot
df = pd.read_excel(r"C:\Users\KRISHNA\Downloads\spread_locator_dataset.xlsx")
df["transaction_date"] = pd.to_datetime(df["transaction_date"])
print(df.head())
print("Dataset Shape:", df.shape)
print(df.info())
print(df.describe())
```

```python
# Count transactions
status_counts = df["transaction_status"].value_counts()
success = status_counts["Success"]
fail = status_counts["Fail"]
total = len(df)
# Probability of success
p = success / total

print("Total Transactions:", total)
print("Successful Transactions:", success)
print("Failed Transactions:", fail)

print("Probability of Success:", p)
print("Probability of Failure:", 1 - p)

# Bernoulli distribution
x = [0, 1]
probabilities = bernoulli.pmf(x, p)

plt.bar(["Fail", "Success"], probabilities)
plt.xlabel("Transaction Status")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution of Transaction Status")
plt.show()
```

```python
#weekly transaction summary
df["week"] = df["transaction_date"].dt.to_period("W")

weekly_data = df.groupby("week").agg(
    total_transactions=("transaction_id", "count"),
    successful_transactions=("transaction_status",
                              lambda x: (x == "Success").sum())
)

print(weekly_data)

#Estimate probability of success
p = (df["transaction_status"] == "Success").mean()

# Binomial distribution for a representative week
n = round(weekly_data["total_transactions"].mean())

x = np.arange(0, n + 1)
binomial_prob = binom.pmf(x, n, p)

plt.figure(figsize=(9, 5))
plt.bar(x, binomial_prob)
plt.xlabel("Number of Successful Transactions")
plt.ylabel("Probability")
plt.title("Binomial Distribution of Weekly Successful Transactions")
plt.show()

print("Average weekly transactions:", weekly_data["total_transactions"].mean())
print("Probability of success:", p)
print("Expected successes in a representative week:", binom.mean(n, p))
```

```python
#Count transactions per day
daily_transactions = df.groupby(
    df["transaction_date"].dt.date
).size()
#Calculate average daily transactions
lambda_value = daily_transactions.mean()
print("Average Transactions Per Day:", lambda_value)
#Fit Poisson distribution
x = np.arange(0, daily_transactions.max() + 1)
poisson_prob = poisson.pmf(x, lambda_value)
#Plot actual and fitted distribution
plt.figure(figsize=(9, 5))
plt.hist(
    daily_transactions,
    bins=np.arange(0, daily_transactions.max() + 2) - 0.5,
    density=True,
    alpha=0.6,
    label="Actual Transactions"
)
plt.plot(
    x,
    poisson_prob,
    "ro-",
    label="Poisson Distribution"
)
plt.xlabel("Number of Transactions Per Day")
plt.ylabel("Probability")
plt.title("Poisson Distribution of Daily Transactions")
plt.legend()
plt.show()
```

```python
#Extract transaction amounts
amounts = df["transaction_amount"].dropna().values
#Fit Log-Normal distribution
shape, loc, scale = lognorm.fit(amounts, floc=0)
print("Shape:", shape)
print("Location:", loc)
print("Scale:", scale)
#Plot actual data and fitted distribution
x = np.linspace(amounts.min(), amounts.max(), 500)
plt.figure(figsize=(9, 5))
sns.histplot(
    amounts,
    bins=30,
    stat="density",
    alpha=0.5,
    label="Actual Transaction Amounts"
)
plt.plot(
    x,
    lognorm.pdf(x, shape, loc, scale),
    "r-",
    linewidth=2,
    label="Log-Normal Fit"
)
plt.xlabel("Transaction Amount (₹)")
plt.ylabel("Density")
plt.title("Log-Normal Distribution of Transaction Amounts")
plt.legend()
plt.show()
```

```python
# Extract transaction amounts
amounts = df["transaction_amount"].dropna().values
# Fit Log-Normal distribution
shape, loc, scale = lognorm.fit(amounts, floc=0)

print("Shape:", shape)
print("Location:", loc)
print("Scale:", scale)
# Plot actual data and fitted distribution
x = np.linspace(amounts.min(), amounts.max(), 500)
plt.figure(figsize=(9, 5))
sns.histplot(
    amounts,
    bins=30,
    stat="density",
    alpha=0.5,
    label="Actual Transaction Amounts"
)
plt.plot(
    x,
    lognorm.pdf(x, shape, loc, scale),
    "r-",
    linewidth=2,
    label="Log-Normal Fit"
)
plt.xlabel("Transaction Amount (₹)")
plt.ylabel("Density")
plt.title("Log-Normal Distribution of Transaction Amounts")
plt.legend()
plt.show()
```

```python
# Select upper 10% transaction amounts
xmin = np.percentile(amounts, 90)
tail_data = amounts[amounts >= xmin]
# Fit Pareto distribution to upper tail
power_shape, power_loc, power_scale = stats.pareto.fit(
    tail_data,
    floc=0
)
print("Minimum value of upper tail:", xmin)
print("Number of high-value transactions:", len(tail_data))
print("Power Law Shape:", power_shape)
# Plot Power Law fit
x_tail = np.linspace(xmin, tail_data.max(), 300)
plt.figure(figsize=(9, 5))

sns.histplot(
    tail_data,
    bins=15,
    stat="density",
    alpha=0.5,
    label="High-Value Transactions"
)

plt.plot(
    x_tail,
    stats.pareto.pdf(
        x_tail,
        power_shape,
        power_loc,
        power_scale
    ),
    "r-",
    label="Power Law / Pareto Fit"
)

plt.xlabel("Transaction Amount (₹)")
plt.ylabel("Density")
plt.title("Power Law Distribution of High-Value Transactions")
plt.legend()
plt.show()
```

```python
# Shapiro-Wilk normality test
statistic, p_value = stats.shapiro(amounts)

print("Test Statistic:", statistic)
print("P-value:", p_value)

if p_value > 0.05:
    print("Data appears approximately normal.")
else:
    print("Data is not normally distributed.")
```

```python
# Apply Box-Cox transformation
transformed_amounts, lambda_value = boxcox(amounts)

print("Box-Cox Lambda:", lambda_value)

# Compare original and transformed data
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(amounts, bins=30)
plt.title("Original Transaction Amounts")
plt.xlabel("Transaction Amount")

plt.subplot(1, 2, 2)
plt.hist(transformed_amounts, bins=30)
plt.title("After Box-Cox Transformation")
plt.xlabel("Transformed Amount")

plt.tight_layout()
plt.show()

print("Original Skewness:", stats.skew(amounts))
print("Transformed Skewness:", stats.skew(transformed_amounts))
```

```python
# Calculate mean and standard deviation
mean_amount = np.mean(amounts)
std_amount = np.std(amounts)

# Calculate Z-scores
df["Z_Score"] = (
    df["transaction_amount"] - mean_amount
) / std_amount

print("Mean Amount:", mean_amount)
print("Standard Deviation:", std_amount)

print(df[[
    "transaction_amount",
    "Z_Score"
]].head())
```

```python
# Calculate probability using fitted Log-Normal distribution
threshold = 5000

probability = lognorm.sf(
    threshold,
    shape,
    loc,
    scale
)

# Calculate actual probability from dataset
actual_probability = np.mean(amounts > threshold)

print("Log-Normal Probability:", probability)
print("Log-Normal Percentage:", probability * 100, "%")

print("Actual Probability:", actual_probability)
print("Actual Percentage:", actual_probability * 100, "%")
```
